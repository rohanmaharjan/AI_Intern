```python
import pandas as pd
import seaborn as sns
import plotly.express as px
import os

os.makedirs("charts", exist_ok=True)

```

```python
# Load seaborn dataset
df = sns.load_dataset("penguins")

# Remove missing values
df = df.dropna()

# Preview dataset
df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
print("Species:", df["species"].unique())
print("Island:", df["island"].unique())
print("Sex:", df["sex"].unique())
```

```python
fig1 = px.scatter(
    df,
    x="bill_length_mm",
    y="bill_depth_mm",
    color="species",
    hover_data=["body_mass_g", "island"],
    title="Interactive Scatter Plot: Bill Length vs Bill Depth",
    size="body_mass_g"
)

fig1.show()
# Export standalone HTML
fig1.write_html("charts/scatter_plot.html")
```

```python
fig2 = px.box(
    df,
    x="species",
    y="flipper_length_mm",
    color="sex",
    hover_data=["body_mass_g", "island"],
    title="Interactive Box Plot: Flipper Length Distribution"
)

fig2.show()

# Export standalone HTML
fig2.write_html("charts/box_plot.html")
```

```python
fig3 = px.histogram(
    df,
    x="body_mass_g",
    color="species",
    hover_data=["island", "sex"],
    nbins=25,
    title="Interactive Histogram: Body Mass Distribution"
)

fig3.show()

# Export standalone HTML
fig3.write_html("charts/histogram_plot.html")
```

```python
treemap_data = df.groupby(["species", "island"]).agg({
    "body_mass_g": "mean",
    "flipper_length_mm": "mean"
}).reset_index()

fig4 = px.treemap(
    treemap_data,
    path=["species", "island"],
    values="body_mass_g",
    color="flipper_length_mm",
    hover_data=["body_mass_g", "flipper_length_mm"],
    title="Treemap: Species and Island Distribution"
)

fig4.show()

# Export standalone HTML
fig4.write_html("charts/treemap_plot.html")
```

```python
print("""
Design Choices Documentation

1. Scatter Plot:
   - Used to explore relationship between bill length and bill depth
   - Bubble size represents body mass
   - Species used as categorical color encoding

2. Box Plot:
   - Highlights distribution and outliers
   - Sex chosen for color comparison
   - Hover adds island and body mass context

3. Histogram:
   - Shows body mass distribution
   - Species-based color segmentation
   - Helps compare frequency patterns

4. Treemap:
   - Advanced chart type
   - Hierarchical grouping by species and island
   - Area shows average body mass
   - Color intensity shows average flipper length
""")
```
