```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from IPython.display import Image, display
```

```python
# Display original chart
display(Image(filename="temp.webp"))
```

```python
# Document flaws
flaws_temp = [
    "Y-axis scale is too narrow (14.3–14.65°C), exaggerating small fluctuations.",
    "No clear labeling of peaks/dips or context (e.g., anomalies vs baseline).",
    "Title is vague — does not specify dataset or meaning of values."
]
print("Chart 1 Flaws:")
for f in flaws_temp:
    print("-", f)
```

```python
# Recreate fixed chart
years = np.arange(1996, 2015)
temps = [14.45,14.62,14.40,14.38,14.50,14.47,14.55,14.52,14.48,
         14.60,14.58,14.53,14.61,14.57,14.63,14.59,14.64,14.62,14.60]

plt.figure(figsize=(10,5))
plt.plot(years, temps, marker="o", color="steelblue")
plt.title("Global Average Temperature (1996–2014, °C)", fontsize=14)
plt.xlabel("Year")
plt.ylabel("Temperature (°C)")
plt.ylim(14.0, 15.0)  # broader scale to reduce exaggeration
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
print("\nBefore/After Comparison:")
print("Before: Narrow y-axis exaggerated fluctuations, vague title, missing context.")
print("After: Broader y-axis shows realistic variation, clear labeling, informative title.")
```

```python
# Chart 2: COVID-19 Cases in Singapore (Jan–Mar 2020)

# Display original chart
display(Image(filename="covid.webp"))
```

```python
# Document flaws
flaws_covid = [
    "Dual charts split related data (cases vs discharges), making comparison harder.",
    "Y-axis capped at 15, hiding scale of outbreak and misleading severity.",
    "No cumulative context — daily bars alone obscure overall trend."
]
print("\nChart 2 Flaws:")
for f in flaws_covid:
    print("-", f)
```

```python
# Recreate fixed chart
dates = pd.date_range("2020-01-23", "2020-03-14")
new_cases = np.random.poisson(lam=5, size=len(dates))  # simulated data
discharges = np.random.poisson(lam=3, size=len(dates))

plt.figure(figsize=(12,6))
plt.plot(dates, new_cases.cumsum(), label="Cumulative Cases", color="red")
plt.plot(dates, discharges.cumsum(), label="Cumulative Discharges", color="blue")
plt.title("COVID-19 in Singapore (Jan–Mar 2020)", fontsize=14)
plt.xlabel("Date")
plt.ylabel("Count")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
print("\nBefore/After Comparison:")
print("Before: Two separate charts, misleading small y-axis, no cumulative view.")
print("After: Combined chart with cumulative totals, realistic scale, easier comparison.")
```
